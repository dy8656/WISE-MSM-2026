The input file is set to be as follows and saved to *lmp.in*

````python
f = open('lmp.in', 'w')
a = """
# PART A - ENERGY MINIMIZATION

# 1) Initialization
units lj               #< units of lj potential
dimension 3            #< 3dimensional box
atom_style atomic
pair_style lj/cut 2.5  #< LJ pair potential with cut-off after 2.5
boundary p p p         #< periodic boundary conditions

# 2) System definition
region simulation_box block -20 20 -20 20 -20 20   #< define a simulation box centered around zero
create_box 2 simulation_box                        #< make the box with 2 different atom types
create_atoms 1 random 5500 921342 simulation_box   #<
create_atoms 2 random 500 225469 simulation_box

# 3) Simulation settings
mass 1 1                                 #< mass of atoms of type 1 is 1
mass 2 1                                 #< mass of atoms of type 2 is 1
pair_coeff 1 1 1.0 1.0                   #< coefficients in LJ potential for interactions of atom of type 1 to type 1
pair_coeff 2 2 0.5 3.0                   #< coefficients in LJ potential for interactions of atom of type 2 to type 2

# 4) Visualization
thermo 10                                #< prints information at each of this steps https://docs.lammps.org/thermo.html
thermo_style custom step etotal press    #< customised output

# 5) Run
minimize 1.0e-6 1.0e-6 1000 10000       #< https://docs.lammps.org/minimize.html, stopping tolerance for energy (unitless), stopping tolerance for force (force units), max iterations of minimizer, max number of force/energy evaluations


# PART B - MOLECULAR DYNAMICS
# 4) Visualization
dump mydmp1 all atom 100 dump.lammpstrj
dump mydmp2 all image 100 dump.*.jpg type type shiny 0.1 box no 0.01 view 0 0 zoom 1.8
dump_modify mydmp2 acolor 1 darkturquoise acolor 2 darkblue adiam 1 1 adiam 2 3 backcolor white
dump myveldmp all custom 100 vel_data.lammpstrj id type vx vy vz
thermo_style custom step temp etotal pe ke press   #< see above
thermo 50                                          #< see above

# radial distribution function
compute RDF all rdf 100 1 1
fix RDF all ave/time 10 10 100 c_RDF[*] file RDF.dat mode vector

# Store thermo variables in seperate variable for printing.
variable K equal ke
variable V equal pe
variable Etot equal etotal
variable myTemp equal temp
variable myStep equal step
variable myPress equal press

# Write specific thermo info to a separate file
fix extra all print 100 "${myStep} ${Etot} ${K} ${V} ${myTemp} ${myPress}" file "./output.txt" screen no

# 5) Run
fix mynve all nve                            #< https://docs.lammps.org/fix_nve.html - Perform plain time integration to update position and velocity for atoms in the group each timestep.
fix mylgv all langevin 1.0 1.0 0.1 1530917   #< Apply a Langevin thermostat, start temp, end temp, damping, seed
timestep 0.005  #< time step for the integration
run 10000       #< run number of steps
write_data output.dat #< prints last configutation
"""
print(a,file=f)
f.close()
````

Although the input file for the structure is provided with this tutorial, here a quick instruction:

### Unbreakable bonds

With most classical molecular dynamics force fields, the chemical bonds between the atoms are set at the start of the simulation. Regardless of the forces applied to the atoms during the simulations, the bonds remain intact. The bonds between neighbor atoms typically consist of springs with given equilibrium distances $r_0$ nd a constant $k_b$ : $U_b=k_b(r-r_0)^2$. Additionally, angular and dihedral constraints are usually applied to maintain the relative orientations of neighbor atoms.

## Using GROMACS

We need to generate a coordinate file first.
The following Python code creates your 6000 atoms with random positions in a cubic box.

````python
import numpy as np

# System parameters
N1 = 5500
N2 = 500
L = 40.0  # nm

N = N1 + N2

positions = np.empty((N, 3))
types = np.empty(N, dtype=int)

n = 0

# Randomly distribute atoms
for atom_type, count, seed, sigma in [
    (1, N1, 921342, 1.0),
    (2, N2, 225469, 3.0)
]:
    rng = np.random.default_rng(seed)

    for _ in range(count):
        for attempt in range(20000):

            pos = rng.uniform(0, L, 3)

            if n > 0:
                dr = positions[:n] - pos

                # Minimum image convention
                dr -= L * np.rint(dr / L)

                other_sigma = np.where(types[:n] == 1, 1.0, 3.0)

                # Reject excessively close atoms
                rmin2 = 0.85**2 * sigma * other_sigma

                if np.any(np.sum(dr**2, axis=1) < rmin2):
                    continue

            positions[n] = pos
            types[n] = atom_type
            n += 1
            break
        else:
            raise RuntimeError("Could not place an atom.")

# Write GROMACS coordinate file
with open("initial.gro", "w") as f:

    f.write("Two-component Lennard-Jones system\n")
    f.write(f"{N:5d}\n")

    for i, (pos, atom_type) in enumerate(zip(positions, types), 1):

        residue = "S1" if atom_type == 1 else "S2"
        atom = "A1" if atom_type == 1 else "A2"

        x, y, z = pos

        f.write(
            f"{i:5d}{residue:<5}{atom:>5}{i:5d}"
            f"{x:8.3f}{y:8.3f}{z:8.3f}\n"
        )

    f.write(f"{L:10.5f}{L:10.5f}{L:10.5f}\n")
````

Compare to the LAMMPS input, the coordinate range is shifted from \([-20,20]\) to \([0,40]\), which makes no physical difference under periodic boundary conditions.

Lets define the Lennard - Jones Potential

This defines two independent monatomic species interacting through

$$
V_{ij}(r)=4\epsilon_{ij}
\left[
\left(\frac{\sigma_{ij}}r\right)^{12}
-
\left(\frac{\sigma_{ij}}r\right)^6
\right].
$$
No external force-field files are needed.

````python
topology = """
; Two-component Lennard-Jones system

[ defaults ]
; nbfunc  comb-rule  gen-pairs  fudgeLJ  fudgeQQ
1         3          no         1.0      1.0

[ atomtypes ]
; name  at.num  mass   charge  ptype  sigma  epsilon
A1      1       1.0    0.0     A      1.0    1.0
A2      1       1.0    0.0     A      3.0    0.5

[ moleculetype ]
; name  nrexcl
Species1  0

[ atoms ]
; nr  type  resnr  residue  atom  cgnr  charge  mass
1     A1    1      S1       A1    1     0.0     1.0

[ moleculetype ]
; name  nrexcl
Species2  0

[ atoms ]
; nr  type  resnr  residue  atom  cgnr  charge  mass
1     A2    1      S2       A2    1     0.0     1.0

[ system ]
Two-component Lennard-Jones mixture

[ molecules ]
Species1  5500
Species2  500
"""

with open("system.top", "w") as f:
    f.write(topology)
````

Energy minimization

````python
minimization = """
; PART A - ENERGY MINIMIZATION

; 1) Initialization
integrator              = steep
nsteps                  = 3000
emtol                   = 10.0
emstep                  = 0.01

; 2) Interaction settings
cutoff-scheme           = Verlet
nstlist                 = 10
verlet-buffer-tolerance = -1
rlist                   = 2.7

vdwtype                 = Cut-off
rvdw                    = 2.5
vdw-modifier            = None
DispCorr                = no

coulombtype             = Cut-off
rcoulomb                = 2.5

; 3) Boundary conditions
pbc                     = xyz

; 4) Output
nstenergy               = 10
nstlog                  = 10
"""

with open("minim.mdp", "w") as f:
    f.write(minimization)
````

GROMACS tests the maximum force on an atom, whereas the LAMMPS minimization uses different stopping criteria, including the global force norm and relative energy changes.

Molecular dynamics

GROMACS's sd integrator combines deterministic forces, friction, and stochastic forces within a single integration scheme.

````python
md = """
; PART B - MOLECULAR DYNAMICS

; 1) Initialization
integrator              = sd
dt                      = 0.005
nsteps                  = 10000

; 2) Interaction settings
cutoff-scheme           = Verlet
nstlist                 = 10
verlet-buffer-tolerance = -1
rlist                   = 2.7

vdwtype                 = Cut-off
rvdw                    = 2.5
vdw-modifier            = None
DispCorr                = no

coulombtype             = Cut-off
rcoulomb                = 2.5

; 3) Boundary conditions
pbc                     = xyz

; 4) Langevin dynamics
tc-grps                 = System
tau-t                   = 0.1
ref-t                   = 120.272
ld-seed                 = 1530917

; Constant volume
pcoupl                  = no

; Initial velocities
gen-vel                 = no

; No periodic removal of COM motion
comm-mode               = None

; 5) Output
nstcalcenergy           = 50
nstenergy               = 50
nstlog                  = 50

; Coordinates and velocities every 100 steps
nstxout                 = 100
nstvout                 = 100

; Compressed trajectory
nstxout-compressed      = 100
"""

with open("md.mdp", "w") as f:
    f.write(md)
````

Last, we need to visualise the band structure. Here, we are using python's matplotlib. The Fermi level is extracted from the output file of the self-consitent calculation

*highest occupied, lowest unoccupied level (ev):     6.2194    6.9241*

````python
import matplotlib.pyplot as plt
from matplotlib import rcParamsDefault
import numpy as np

plt.rcParams["figure.dpi"]=150
plt.rcParams["figure.facecolor"]="white"
plt.rcParams["figure.figsize"]=(4, 6)

data = np.loadtxt('si_bands.dat.gnu') #< load data

k = np.unique(data[:, 0]) #< copy the data to k-points
bands = np.reshape(data[:, 1], (-1, len(k))) #< < copy the data to bands

ef = 6.57235 #< Fermi energy

#< plot the bands
for band in range(len(bands)):
    plt.plot(k, bands[band, :]-ef, linewidth=1, alpha=0.5, color='k')

#< axes limits
plt.xlim(min(k), max(k))
plt.ylim(-20,10)

#< draw horizontal line at the Fermi energy
plt.axhline(0.0, linestyle=(0, (5, 5)), linewidth=0.75, color='k', alpha=0.5)

#< draw vertical lines at high symmetry k-points (check bands_pp.out)
plt.axvline(0.8660, linewidth=0.75, color='k', alpha=0.5)
plt.axvline(1.8660, linewidth=0.75, color='k', alpha=0.5)
plt.axvline(2.2196, linewidth=0.75, color='k', alpha=0.5)

#< put text labels on the high symmetry k-points
plt.xticks(ticks= [0.0000, 0.8660, 1.8660, 2.2196, 3.2802], \
labels=['L', '$\Gamma$', 'X', 'U', '$\Gamma$'])

#< y-label
plt.ylabel("$E-E_F$ (eV)")

#< save and visualize figure
plt.tight_layout()
plt.savefig('bands.pdf')
plt.show()
````
